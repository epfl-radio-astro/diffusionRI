```python
import numpy as np

mask = np.load('../uv_coverages/uv_sampling_60_steps_centered.npy')
N = mask.shape[0]
```

```python
img = np.random.default_rng(0).standard_normal((N, N))
F = np.fft.fftshift(np.fft.fft2(img))
Fs = F * mask
out = np.fft.ifft2(np.fft.ifftshift(Fs))

max_im = np.abs(out.imag).max()
max_re = np.abs(out.real).max()
print(max_im, max_re, max_im / max_re)
assert max_im < 1e-7
print('PASS')
```
